```python
import pandas as pd
```

```python
import glob
```

```python
files = glob.glob('raw/*.xls')
```

```python
print(len(files))
```

```python
def parse_filename(filepath):
    name = filepath.split('\\')[-1].replace('.xls', '')
    state_code, year = name.split(' ')
    return state_code, year
```

```python
# 3 - Processing one file - clean harder, sum real districts
```

```python
def process_file(filepath):
    state_code, year = parse_filename(filepath)
    df = pd.read_html(filepath)[0]
    df.columns = df.iloc[1]
    df = df.iloc[5:]
    
    districts = df[pd.to_numeric(df['S.No'], errors='coerce').notna()]
    util_pos = df.columns.get_loc('%age Utilization')
    
    availability = pd.to_numeric(districts['Total Availabilty'], errors='coerce').sum()
    expenditure = pd.to_numeric(districts.iloc[:, util_pos - 1], errors='coerce').sum()
    
    return {'state': state_code, 'year': year, 'availability': availability, 'expenditure': expenditure}
```

```python
results = []
errors = []

for f in files:
    try:
        results.append(process_file(f))
    except Exception as e:
        errors.append((f, str(e)))

print(len(results), 'succeeded')
print(len(errors), 'failed')
print(errors[:5])
```

```python
# 5 - FINAL TABLE
```

```python
df_final = pd.DataFrame(results)
```

```python
df_final['utilization_rate'] = df_final['expenditure'] / df_final['availability'] * 100
```

```python
import sqlite3
```

```python
conn = sqlite3.connect('mgnrega.db')
```

```python
df_final.to_sql('utilization', conn, if_exists = 'replace', index = False)
```

```python
pd.read_sql('SELECT * FROM utilization LIMIT 5', conn)
```

```python
#National average utilization by year
```

```python
query1 = """
SELECT year, AVG(utilization_rate) as national_avg
FROM utilization
WHERE state != 'WB'
GROUP BY year
ORDER BY year
"""
```

```python
pd.read_sql(query1, conn)
```

```python
# BOTTOM 5 STATES PER YEAR
```

```python
query2 = """
SELECT state, year, utilization_rate, RANK() OVER (PARTITION BY year ORDER BY utilization_rate ASC) as rank_in_year
FROM utilization
WHERE state != 'WB'
"""
```

```python
ranked = pd.read_sql(query2, conn)
```

```python
ranked[ranked['rank_in_year'] <= 5].sort_values(['year', 'rank_in_year'])
```

```python
# Year by Year change per state
```

```python
query3 = """
SELECT state, year, utilization_rate,
utilization_rate - LAG(utilization_rate) OVER (PARTITION BY state ORDER BY year) as yoy_change
FROM utilization
WHERE state != 'WB'
ORDER BY state, year
"""
```

```python
pd.read_sql(query3, conn).head(12)
```
